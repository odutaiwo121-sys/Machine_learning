```python
sales = [
    {"month": "january", "sales": 100000},
    {"month": "february", "sales": 200000},
    {"month": "march", "sales": 300000},
    {"month": "april", "sales": 400000},
]

for month in sales:
    print(month)
```

```python
hightest_month = sales[0]

for month in sales:
    if month["sales"] > highest_month["sales"]:
        highest_month = month

print(highest_month)        
```
